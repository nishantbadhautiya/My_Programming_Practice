```python
import cv2
```

```python
cap = cv2.VideoCapture(0) # Open webcam
```

```python
if not cap.isOpened(): 
    print("Error: Cannot open the camera")
    exit()
```

```python
cap.isOpened()
```

### 1. Capture Video from Webcam

```python
while True: 
    success, frame = cap.read() # read the frame 
    if not success: 
        print("Can not read the frame")
    cv2.imshow(winname="VIDEO", mat=frame) 
    # Break the loop on pressing 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release() 
cv2.destroyAllWindows() 
```

### 2. Capture Video from a File

```python
cap = cv2.VideoCapture('output_video.mp4')
```

### 3. Get Video Properties

```python
# Get properties
width = cap.get(propId=cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(propId=cv2.CAP_PROP_FRAME_HEIGHT)
fps = cap.get(propId=cv2.CAP_PROP_FPS)
frame_count = cap.get(propId=cv2.CAP_PROP_FRAME_COUNT)

print(f"Width: {width}, Height: {height}, FPS: {fps}, Total Frames: {frame_count}")
```

### 4. Setting Video Properties

```python
cap = cv2.VideoCapture(0)
# Set resolution
cap.set(propId=cv2.CAP_PROP_FRAME_WIDTH, value=1280)
cap.set(propId=cv2.CAP_PROP_FRAME_HEIGHT, value=720)
```

```python
while True:
    ret, frame = cap.read()
    if not ret:  # If video ends, exit
        break

    cv2.imshow("Video Playback", frame)

    if cv2.waitKey(30) & 0xFF == ord('q'):  # Play video at 30 fps
        break

cap.release()
cv2.destroyAllWindows()
```

# 5. Processing Frames

### 5.1 Convert Frames to Grayscale

```python
cap = cv2.VideoCapture(0)
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray_frame = cv2.cvtColor(src=frame, code=cv2.COLOR_BGR2GRAY)
    cv2.imshow("Grayscale Video", gray_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

### 5.2 Edge Detection on Frames

```python
cap = cv2.VideoCapture(0)
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Apply edge detection
    edges = cv2.Canny(image=frame, threshold1=100, threshold2=200) # most commonly used edge detection method is the Canny Edge Detection algorithm.

    # Display edges
    cv2.imshow("Edges in Video", edges)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release() 
cv2.destroyAllWindows() 
```

### 6. Adding Overlays

```python
cap = cv2.VideoCapture(0) 
```

```python
# You can draw shapes or add text to frames in real time.
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Draw a rectangle
    cv2.rectangle(img=frame, pt1=(50, 50), pt2=(200, 200), color=(0, 255, 0), thickness=2)

    # Add text
    cv2.putText(frame, "Hello OpenCV!", (50, 300), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    # Show the frame
    cv2.imshow("Video with Overlay", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release() 
cv2.destroyAllWindows() 
```

### 6. Extracting Frames from Video

```python
cap = cv2.VideoCapture(0) 
```

```python
frame_number = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Save frames to disk
    cv2.imwrite(f"frame_{frame_number}.jpg", frame)
    frame_number += 1

    cv2.imshow("Frame Saving", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release() 
cv2.destroyAllWindows() 
```

### 7. Skipping Frames

```python
cap = cv2.VideoCapture(0) 
```

```python
# Sometimes, you might want to skip frames to reduce processing.
frame_skip = 5
current_frame = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Process every nth frame
    if current_frame % frame_skip == 0:
        cv2.imshow("Processed Frame", frame)

    current_frame += 1

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release() 
cv2.destroyAllWindows() 
```

### 8. Background Subtraction

```python
cap = cv2.VideoCapture(0) 
```

```python
# You can extract moving objects by subtracting the background using cv2.createBackgroundSubtractorMOG2.
# Create the background subtractor
fgbg = cv2.createBackgroundSubtractorMOG2()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Apply background subtraction
    fgmask = fgbg.apply(frame)

    # Show the foreground mask
    cv2.imshow("Foreground Mask", fgmask)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release() 
cv2.destroyAllWindows() 
```

### 9. Face Detection (Haar Cascades)

```python
cap = cv2.VideoCapture(0) 
```

```python
# Load Haar cascade for face detection
# Using pre-trained Haar cascades for face detection:
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
```

```python
while True:
    # Capture each frame
    ret, frame = cap.read()
    if not ret:
        print("Video ended or failed to capture frame.")
        break

    # Convert frame to grayscale for better face detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces in the frame
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    # Draw rectangles around detected faces
    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)

    # Display the frame with rectangles drawn
    cv2.imshow("Face Detection", frame)

    # Break the loop if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources and close windows
cap.release()
cv2.destroyAllWindows()
```
