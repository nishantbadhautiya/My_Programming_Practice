```python
import cv2
```

### 1. Reading the image

```python
image = cv2.imread('1_image.png') # Read an image
```

### 2. Accessing Image Properties

```python
# We can extract various properties of an image like its dimensions, number of channels, etc.
height, width, channels = image.shape
print(f"Width: {width}, Height: {height}, Channels: {channels}")

# Access pixel value at (x=100, y=50)
pixel_value = image[50, 100]  # Returns a tuple (B, G, R)
print(f"Pixel Value at (50, 100): {pixel_value}")
# Modify pixel value
image[50, 100] = (0, 255, 0)  # Set it to green
```

### 3. Converting Color Spaces

```python
# OpenCV allows you to convert images to different color spaces. some color formats are 
# (1) BGR: Default color format.
# (2) Grayscale: Single-channel image (black and white).

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) # Convert BGR to Grayscale
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV) # Convert BGR to HSV (Hue, Saturation, Value)
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # Convert BGR to RGB
```

### 4. Resizing Images

```python
# Resizing changes the width and height of an image. This is useful for scaling images up or down.
resized_image = cv2.resize(src=image, dsize=(400, 400)) # Resize to 400x400 pixels

# Resize with scaling factors
scaled_image = cv2.resize(src=image, dsize=None, fx=0.8, fy=0.8) # Scale down by 50%
```

### 5. Cropping an Image

```python
image.shape
```

```python
# Cropping extracts a specific region of an image using slicing.
# Crop a region of interest (ROI)
cropped_image = image[50:200, 100:300]# y1:y2, x1:x2
```

### 6. Rotating Images

```python
# Images can be rotated using rotation matrices.
rotated_image = cv2.rotate(src=image, rotateCode=cv2.ROTATE_90_CLOCKWISE)

# Custom Rotation 
(h, w) = image.shape[:2]
center = (w // 2, h / 2) 
rotation_matrix = cv2.getRotationMatrix2D(center=center, angle=-45, scale=1.0) 
print(rotation_matrix) 
rotated_custom = cv2.warpAffine(src=image, M= rotation_matrix, dsize=(w, h)) # warpAffine applies the affine transformation
```

### 7. Flipping an Image

```python
# Flipping allows you to reverse the orientation of an image.
flipped_horizontal = cv2.flip(src=image, flipCode=1) # Flip horizontally
flipped_vertica = cv2.flip(src=image, flipCode=0) # Flip vertically 
flipped_both = cv2.flip(src=image, flipCode=-1) # # Flip both horizontally and vertically
```

### 8. Drawing Shapes

```python
cv2.line(img=resized_image, pt1=(50, 50), pt2=(200, 200), color=(255, 0, 0), thickness=3) # Draw a line of BGR color, thickness 3
cv2.rectangle(img=image, pt1=(100, 100), pt2=(300, 300), color=(255, 255, 0), thickness=3) # Draw a rectangle
cv2.circle(img=image, center=(250, 250), radius=120, color=(0, 255, 0), thickness=-1) # -1 fills the circle

# Draw text
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img=image, text="Hello OpenCV! ", org=(20, 100), fontFace=font, fontScale=1, color=(255, 255, 0), thickness=2) # org stands for "origin". It specifies the coordinates of the starting point (x, y) where the text will be drawn. in this case x = 20: 20 pixels from the left edge. and y = 100: 100 pixels from the top edge (the baseline of the text starts here).
```

### 9. Image Blending

```python
# Load two images of the same size
image1 = cv2.imread("1_image.png")
image2 = cv2.imread("2_image.png")
image2 = cv2.resize(src=image2, dsize=(image1.shape[1], image1.shape[0])) # To ensure both images have the same dimensions
# Blend with weights
blended = cv2.addWeighted(src1=image1, alpha=0.7, src2=image2, beta=0.3, gamma=0)  # 70% of image1, 30% of image2

#image1 and image2: Input images.
# alpha and beta: The weights applied to each image (how much of each image to blend).
# gamma: Scalar added to each sum (used to adjust brightness).
# Result:
#     output pixel=image1_pixel⋅α+image2_pixel⋅β+γ
#     output pixel=image1_pixel⋅α+image2_pixel⋅β+γ
```

```python
print(image1.shape)
print(image2.shape)
```

### 10. Image Arithmetic

```python
# You can perform addition, subtraction, etc., directly on images.
added_image = cv2.add(src1=image1, src2=image2) # Add two images
subtracted_image = cv2.subtract(image1, image2) # Subtract one image from another
```

```python
subtracted_image.shape
```

### 11. Splitting and Merging Channels

```python
# You can split an image into its component channels or merge separate channels into one image.
b, g, r = cv2.split(m=image1) 
merged_image = cv2.merge((b, g, r)) # Merge channels
```

```python
merged_image.shape
```

### 13. Gaussian Blur

```python
# Smooth an image using Gaussian blur.
# Apply Gaussian blur
blurred_image = cv2.GaussianBlur(src=image1, ksize=(15, 15), sigmaX=0) # Kernel size = (15, 15) Larger kernel sizes result in more blurring sigmaX: Standard deviation of the Gaussian in the X-direction.
```

### 15. Saving the Processed Image

```python
cv2.imwrite(filename="output.png", img=blurred_image)
```
