```python
import cv2
import mediapipe as mp
import time
```

```python
# cap = cv2.VideoCapture("justin_song.webm")
cap = cv2.VideoCapture(0)
pTime = 0

mpFaceDetection = mp.solutions.face_detection
mpDraw = mp.solutions.drawing_utils
faceDetection = mpFaceDetection.FaceDetection(0.75) # Initializes MediaPipe face detection with a confidence threshold of 0.75.
```

```python
def fancyDraw(img, img_shape, bbox): 
    ih, iw, ic = img.shape # 480 640 3
    # (231, 216, 199, 199)
    xmin, ymin, width, height = bbox 
    cv2.line(img=img, pt1=(xmin, ymin), pt2=(xmin+30, ymin), color=(255, 0, 0), thickness=5) 
    cv2.line(img=img, pt1=(xmin, ymin), pt2=(xmin, ymin+30), color=(255, 0, 0), thickness=5) 
    
    cv2.line(img=img, pt1=(xmin+width, ymin), pt2=(xmin+width-30, ymin), color=(255, 0, 0), thickness=5) 
    cv2.line(img=img, pt1=(xmin+width, ymin), pt2=(xmin+width, ymin+30), color=(255, 0, 0), thickness=5)

    cv2.line(img=img, pt1=(xmin, ymin+height), pt2=(xmin, ymin+height-30), color=(255, 0, 0), thickness=5) 
    cv2.line(img=img, pt1=(xmin, ymin+height), pt2=(xmin+30, ymin+height), color=(255, 0, 0), thickness=5)

    cv2.line(img=img, pt1=(xmin+width, ymin+height), pt2=(xmin+width-30, ymin+height), color=(255, 0, 0), thickness=5) 
    cv2.line(img=img, pt1=(xmin+height, ymin+height), pt2=(xmin+width, ymin+height-30), color=(255, 0, 0), thickness=5) 
```

```python
while True:
    success, img = cap.read()

    # img = cv2.resize(img, (640, 860))
    imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    results = faceDetection.process(imgRGB)

    if results.detections:
        for id, detection in enumerate(results.detections):
            # mpDraw.draw_detection(img, detection)
            # print(id, detection)
            # print(detection.score)
            # print(detection.location_data.relative_bounding_box)
            bboxC = detection.location_data.relative_bounding_box
            ih, iw, ic = img.shape # 480 640 3
            bbox = int(bboxC.xmin * iw), int(bboxC.ymin * ih), int(bboxC.width * iw), int(bboxC.height * ih)
            for id2, lm in enumerate(detection.location_data.relative_keypoints):
                if id2 == 0: 
                    cx, cy = int(lm.x * iw), int(lm.y * ih) 
                    cv2.circle(img, (cx, cy), 10, (255, 0, 0),-1) 

            cv2.rectangle(img, bbox, (255, 0, 255), 2)
            fancyDraw(img, img.shape, bbox)
            cv2.putText(img, f'{int(detection.score[0] * 100)}%',(bbox[0], bbox[1] - 20), 
                        cv2.FONT_HERSHEY_PLAIN,2, (255, 0, 255), 2)

    cTime = time.time()
    fps = 1 / (cTime - pTime)
    pTime = cTime
    cv2.putText(img, f'FPS: {int(fps)}', (20, 70), cv2.FONT_HERSHEY_PLAIN,3 , (0, 255, 0), 2)
    cv2.imshow("Image", img)
    if cv2.waitKey(20) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
